```python
import pandas as pd
import numpy as np

# Target
Y = data["Alzheimer’s Diagnosis"]

# Features
X = data.drop(columns=[
    "Subject ID",
    "MRI ID",
    "Alzheimer’s Diagnosis",
    "Hand"
])

# Fill missing values
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy="median")
X = imputer.fit_transform(X)

# Scale
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X = scaler.fit_transform(X)

# Split
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.30, random_state=42
)

# Verify there are no NaNs
print(np.isnan(X_train).sum())
```

```python
#Train the model
```

```python
from sklearn.linear_model import LogisticRegression

#create the Logistic Regression Model
lr = LogisticRegression()

#train the model based on data
lr.fit(X_train, Y_train)

#predict the target variable from data
Y_Predictions = lr.predict(X_test)
```

```python
#Balancing the weights of the different risk factors
lr = LogisticRegression(class_weight="balanced")
lr.fit(X_train, Y_train)
```

```python
Y_prob = lr.predict_proba(X_test)[:,1]

Y_Predictions = (Y_prob >= 0.4).astype(int)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(Y_test, Y_Predictions)
print(f"Accuracy: {accuracy: .2f}")
```

```python
from sklearn.metrics import classification_report

print(classification_report(Y_test, Y_Predictions))
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, accuracy_score

# Create the XGBoost model
xgb = XGBClassifier(
    n_estimators=200,
    max_depth=5,
    learning_rate=0.1,
    random_state=42,
    eval_metric="logloss"
)

# Train the model
xgb.fit(X_train, Y_train)

# Make predictions
Y_pred_xgb = xgb.predict(X_test)

# Evaluate
print("Accuracy:", accuracy_score(Y_test, Y_pred_xgb))
print(classification_report(Y_test, Y_pred_xgb))
```

```python
import numpy as np
from scipy.stats import pointbiserialr, ttest_ind

# SES is Feature 5
ses = X[:, 5]

# Check diagnosis categories
print("Diagnosis values:", np.unique(Y, return_counts=True))

# Make sure diagnosis is binary
groups = np.unique(Y)

for group in groups:
    print(
        f"Group {group}:",
        "n =", np.sum(Y == group),
        "mean SES =", np.mean(ses[Y == group])
    )

# Statistical tests
group1 = ses[Y == groups[0]]
group2 = ses[Y == groups[1]]

t_stat, p_value = ttest_ind(group1, group2, equal_var=False)

correlation, correlation_p = pointbiserialr(Y, ses)

print("\nResults:")
print("t-test p-value:", p_value)
print("Point-biserial correlation:", correlation)
print("Correlation p-value:", correlation_p)
```
