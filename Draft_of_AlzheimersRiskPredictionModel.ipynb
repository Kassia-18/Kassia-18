```python
## Setting the Yes and No in the Diagnosis to 1 and 0 in order to make it compatible
data['Alzheimer’s Diagnosis'] = [1 if value == "Yes" else 0 for value in data['Alzheimer’s Diagnosis']]
```

```python
#Making it a category as well instead of just integer
data['Alzheimer’s Diagnosis'] = data['Alzheimer’s Diagnosis'].astype("category", copy=False)
```

```python
#Looking at the data to confirm
data.head()
```

```python
# divide into target variable and predictors
Y = data['Alzheimer’s Diagnosis']
X = data.drop(['Alzheimer’s Diagnosis'], axis=1)
```

```python
data['Alzheimer’s Diagnosis']
```

```python
X.dtypes
```

```python
X = pd.get_dummies(X, drop_first=True)
```

```python
from sklearn.preprocessing import StandardScaler

# create a scaler object
scaler = StandardScaler()

# fit the scaler to the data and transform the data
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
#split the data
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X_scaled, Y, test_size=0.30, random_state=42)
```

```python
#Train the model
```

```python
from sklearn.linear_model import LogisticRegression

#create the Logistic Regression Model
lr = LogisticRegression()

#train the model based on data
lr.fit(X_train, Y_train)

#predict the target variable from data
Y_Predictions = lr.predict(X_test)
```

```python
#Balancing the weights of the different risk factors
lr = LogisticRegression(class_weight="balanced")
lr.fit(X_train, Y_train)
```

```python
Y_prob = lr.predict_proba(X_test)[:,1]

Y_Predictions = (Y_prob >= 0.4).astype(int)
```

```python
Y_Predictions
```

```python
Y_test
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(Y_test, Y_Predictions)
print(f"Accuracy: {accuracy: .2f}")
```

```python
from sklearn.metrics import classification_report

print(classification_report(Y_test, Y_Predictions))
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, accuracy_score

# Create the XGBoost model
xgb = XGBClassifier(
    n_estimators=200,
    max_depth=5,
    learning_rate=0.1,
    random_state=42,
    eval_metric="logloss"
)

# Train the model
xgb.fit(X_train, Y_train)

# Make predictions
Y_pred_xgb = xgb.predict(X_test)

# Evaluate
print("Accuracy:", accuracy_score(Y_test, Y_pred_xgb))
print(classification_report(Y_test, Y_pred_xgb))
```
